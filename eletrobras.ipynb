```python
import yfinance as yf
import pandas as pd

# Tickers antigo e novo
tickers = ["ELET6.SA", "AXIA6.SA"]

# Baixar dados históricos
dfs = {}
for t in tickers:
    data = yf.download(t, progress=False)
    dfs[t] = data
    print(f"\nTicker: {t}")
    print(f"Período disponível: {data.index.min().date()} até {data.index.max().date()}")
    print(f"Total de registros: {len(data)}")

# Último registro de ELET6 e primeiro de AXIA6
elet_last = dfs["ELET6.SA"].iloc[-1]
axia_first = dfs["AXIA6.SA"].iloc[0]

# Garante extração correta mesmo se vier MultiIndex
elet_close = float(elet_last["Close"]) if "Close" in elet_last else float(elet_last.loc["Close"])
axia_close = float(axia_first["Close"]) if "Close" in axia_first else float(axia_first.loc["Close"])

print("\n--- Comparação de transição ---")
print(f"Último dia ELET6: {dfs['ELET6.SA'].index[-1].date()}, Fechamento: {elet_close:.2f}")
print(f"Primeiro dia AXIA6: {dfs['AXIA6.SA'].index[0].date()}, Fechamento: {axia_close:.2f}")

# Verificar se há lacuna de datas
diff_days = (dfs["AXIA6.SA"].index[0] - dfs["ELET6.SA"].index[-1]).days
if diff_days <= 3:
    print(f"\n✅ As séries parecem contínuas (lacuna de {diff_days} dias).")
else:
    print(f"\n⚠️ Há uma lacuna de {diff_days} dias entre as séries.")
```
