```python
import yfinance as yf
import pandas as pd

# Lista de tickers
tickersus = ['^GSPC','USDBRL=X','AAPL','AIG','BAC','RIO','DHI','EXC','KMB',
              'KO','LOPE','LYB','MGA','MSFT','MSTR','NUE','NVDA','TGT',
              'TMUS','UPS','UNH','XOM']

# Dicionário para armazenar os dados
dados = {}

for t in tickersus:
    try:
        ticker = yf.Ticker(t)
        info = ticker.info  # retorna todas as variáveis disponíveis
        dados[t] = info
        print(f"Coletado: {t}")
    except Exception as e:
        print(f"Erro ao coletar {t}: {e}")

# Converter para DataFrame
df = pd.DataFrame(dados).T  # transpor: linhas = tickers, colunas = variáveis

# Exibir tamanho e primeiras colunas
print(f"\nTotal de colunas: {len(df.columns)}")
print(df.head())
display(df)

# (Opcional) Salvar em Excel para análise
# df.to_excel("dados_yahoo_us_stocks.xlsx")
```
